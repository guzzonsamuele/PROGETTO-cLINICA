```python
from jinja2 import Environment, FileSystemLoader
import os

# 1. Configura l'ambiente per caricare i file dalla cartella corrente
file_loader = FileSystemLoader('.')
env = Environment(loader=file_loader)

# 2. Carica il tuo file XML (assicurati che il nome sia identico)
template = env.get_template('clinical_document.xml')

# 3. Definisci i dati (devono corrispondere ai nomi tra le {{ }})
dati = {
    "data_creazione": "20260409143000",
    "versione": "1",
    "id_documento": "REP_001" # Aggiungilo se lo inserisci nel template
}

# 4. Genera l'XML finale
output = template.render(dati)

# 5. Salva il file sul computer
with open("CDA_Progetto_Team4.xml", "w", encoding="utf-8") as f:
    f.write(output)
print(output)
print("File XML generato con successo!")
```
