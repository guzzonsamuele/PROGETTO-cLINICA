```python
import os
from datetime import datetime
from jinja2 import Environment, FileSystemLoader

# 1. Configura Jinja2 per dire dove trovare i template (nella stessa cartella)
file_loader = FileSystemLoader('PARTE_SG')
env = Environment(loader=file_loader)

# Carica il file XML che hai appena creato
template = env.get_template('clinical_document.xml')

# 2. Prepara i dati fittizi del paziente (in futuro questi arriveranno dal vostro dataset)
dati_paziente = {
    "data_referto": datetime.now().strftime("%Y%m%d%H%M%S"), # Formato HL7: YYYYMMDDHHMMSS
    "id_paziente": "PD-100", # Paziente con Parkinson dal dataset
    "nome_paziente": "Giuseppe",
    "cognome_paziente": "Verdi",
    "sesso": "M",
    "data_nascita": "19550315" # 15 Marzo 1955
}

# 3. Fai il rendering del template passandogli i dati
xml_generato = template.render(dati_paziente)

# 4. Salva il risultato in un nuovo file
nome_file_uscita = f"referto_televisita_{dati_paziente['id_paziente']}.xml"

with open(nome_file_uscita, "w", encoding="utf-8") as file_out:
    file_out.write(xml_generato)

print(f"Successo! File generato e salvato come: {nome_file_uscita}") 
print(xml_generato)
```
